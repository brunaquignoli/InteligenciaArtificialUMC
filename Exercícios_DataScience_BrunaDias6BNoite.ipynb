# Exercícios de SciPy

##  1. Calcule a integral da função f(x) = x2 sin(x) no intervalo de 0 a π.

```python
import numpy as np
import scipy as sp
import scipy.fft as fft
from scipy import integrate
from scipy.integrate import solve_ivp
from scipy.optimize import root, minimize

def f(x):
    return (x ** 2) * np.sin(x)

resultado, erro = integrate.quad(f, 0, np.pi)
print(f"resultado: {resultado:.2f}")
print(f"erro: {erro}")
```

aqui eu criei a função para calcular f(x) com a fórmula passada no exercício e utilizei integrate.quad para fazer o calculo a partir da funcao utilizando np.pi, exibindo o valor de resultado e de erro no final.

## 2. Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10

```python
def eq_diferencial(t, y):
    return t - y
solucao = solve_ivp(eq_diferencial, [0, 10], [1] , t_eval = np.linspace(0 , 10, 100))
print(f"tempo: {solucao.t}")
print(f"\n")
print(f"solucao para y: {solucao.y[0]}")
```

fiz uma função para retornar t-y e depois utilizei solve_ivp para fazer o diferencial diretamente no intervalo entre o e 10, printando a solução de tempo e para y, no final.

## 3. Encontre as raízes da função f(x) = x4 − 3x3 + 2 utilizando SciPy.

```python
def f(x):
    return x**4 - 3* x **3 + 2
solucao = root(f, 0)
print(f"Raiz: {solucao.x}")
```

## 4. Use a função de otimização de SciPy para maximizar f(x) = −x2+4x+1.

```python
def f(x):
    return -x**2 + 4*x + 1

resultado = minimize(lambda x: -f(x), 0)
print(f"Máximo encontrado em x={resultado.x}, f(x)={f(resultado.x)}")
```

utilizei a função como pedido no enunciado, fiz o resultado a partir do lambda para inverter o resultado de minimize (já que não existe um maximize) e printei os resutlados.

## 5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
t = np.linspace(0, 1, 500)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
frequencias = np.fft.fft(sinal)
print(f"Transformada de fourier: {frequencias}")
```

utilizei np.linspace para gerar os números e fiz a conta a partir da formula, salvando o resultado dentro de sinal, então utilizei np.fft para pegar o resultado e passar para dentro de uma variável, printando o resultado.
