```python
array = np.random.rand(10)
print(array)
array = (array - np.min(array)) / (np.max(array) - np.min(array))
print(array)
```

11. Gere um array de 100 números inteiros e encontre o valor que mais se
repete no array.


```python
array = np.random.randint(0, 100, 100)
valores, contagens = np.unique(array, return_counts=True)
frequente = valores[np.max(contagens)]

print(f"o valor mais frequente é {frequente}")
```
