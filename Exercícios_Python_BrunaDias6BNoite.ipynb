21. Busca Binária: Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python
lista = [12378, 400, 325, 20, 21, 19, 6070, 801, 1000, 2, 20]
alvo = int(input("Digite o número que deseja procurar: "))

inicio = 0
fim = len(lista) - 1
meio = 0
achou = False

while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == alvo:
        achou = True
        break
    elif lista[meio] < alvo:
        inicio = meio + 1
    else:
        fim = meio - 1

if achou:
    print(f"O número {alvo} foi encontrado na posição {meio}.")
else:
    print(f"O número {alvo} não foi encontrado.")
```

22. Criptografia de Cifra de César: Crie uma função para criptografar uma mensagem usando a cifra de
César.

```python
mensagem = input("Digite a mensagem para criptografar: ") # pelo que pesquisei sempre tem uma chave mas não sei se é necessário pedir, então vou deixar ela já definida aqui em 2
tamanho = len(mensagem)
chave = 2
base = 'abcdefghijklmnopqrstuvwxyz' # precisa do alfabeto pra usar de base pra mudança
resultado = ''

for i in range(tamanho): # pegando com base no tamanho da entrada
    if mensagem[i] in base:
        posicao = base.index(mensagem[i]) # mover o index com base na posição atual
        nova_posicao = (posicao + chave) % 26 # 26 das letras do alfabeto, mudando a posicao com base na chave definida
        resultado += base[nova_posicao]
    else:
        resultado += mensagem[i] # soma as palavras porque se nao fica tudo junto

print(f"Mensagem criptografada: {resultado}")

```

30. Jogo da Forca: Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.

```python
palavra = "spongebob"
letras_descobertas = []
tentativas = 6

while tentativas > 0:
    print("\npalavra:", end=" ")
    for letra in palavra:
        if letra in letras_descobertas:
            print(letra, end=" ")
        else:
            print("x", end=" ")
    print()

    chute = input("Digite uma letra: ").lower()

    if chute in letras_descobertas:
        print("já foi")
        continue

    if chute in palavra:
        letras_descobertas.append(chute)
        print("Acertou!!")
    else:
        tentativas -= 1
        print(f"errou! Tentativas restantes: {tentativas}")

    if all(letra in letras_descobertas for letra in palavra):
        print(f"Certo! a palavra é {palavra}")
        print("Você venceu!")
        break

if tentativas == 0 and not all(letra in letras_descobertas for letra in palavra):
    print(f"Você perdeu! a palavra era: {palavra}")
```

31. Ordenação por Bolha: Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
numeros = [12378, 400, 325, 20, 21, 19, 6070, 801, 1000, 2, 20]
n = len(numeros)

for i in range(n):
    for j in range(0, n - i - 1):
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print(f"Lista ordenada:  {numeros}")
```

37. Funções Recursivas: Implemente uma função recursiva para calcular o fatorial de um
número.

```python
number = int(input("de qual número devemos calcular fatorial? "))

def fatorial(num):
    resultado = 1
    formatacao = " "

    for i in range(1, num + 1):
        resultado *= i

    formatNum = num
    for i in range(1, formatNum + 1):
        if formatNum != 1:
            formatacao += f" {formatNum} *"
            formatNum -= 1
        else:
            formatacao += f" {formatNum}"

    return resultado, formatacao.strip()

resultado, formatacao = fatorial(number)

print(f"O fatorial de {number} é {resultado} a partir de:\n{number}! = {formatacao}")
```

41. Remover Duplicatas: Remova duplicatas de uma lista de números.


```python
def remover_duplicatas(lista):
    sem_duplicatas = []
    for n in lista:
        if n not in sem_duplicatas:
            sem_duplicatas.append(n)
    return sem_duplicatas

numeros = [12378, 325, 400, 2, 325, 20, 21, 19, 6070, 801, 1000, 2, 20]
print(f"retirando duplicatas: {remover_duplicatas(numeros)}")
```

```python
n = int(input("Insira um número: ")) #pedindo numero

def primos_ate(n):
  primos = []
  for numero in range(2, n + 1):
      primo = True
      for divisor in range(2, numero):
          if numero % divisor == 0:
              primo = False
              break
      if primo:
          primos.append(numero)
  return primos

print(f"os primos são: {primos_ate(n)}" )
```

44. Fibonacci em Nível: Gere a sequência de Fibonacci até um determinado número de
termos.


```python
number = int(input("Digite quantos termos: "))
def fibonacci_termos(number):
    a, b = 0, 1
    sequencia = []

    for _ in range(number):
        sequencia.append(a)
        a, b = b, a + b

    return sequencia

print(fibonacci_termos(number))
```

45. Par ou Ímpar em Lista: Classifique os números de uma lista como pares ou ímpares.

```python
numeros = [12378, 400, 325, 20, 21, 19, 6070, 801, 1000, 2, 20]
def separar(lista):
    pares = []
    impares = []

    for n in lista:
        if n % 2 == 0:
            pares.append(n)
        else:
            impares.append(n)

    return pares, impares

pares, impares = separar(numeros)
print(f"pares: {pares}")
print(f"impares: {impares}")
```

46. Calcular Média e Desvio Padrão: Calcule a média e o desvio padrão de uma lista de números.

```python
def media_e_desvio(lista):
    media = sum(lista) / len(lista)
    variancia = 0

    for n in lista:
        variancia += (n - media) ** 2

    variancia /= len(lista)
    desvio = variancia ** 0.5
    return media, desvio

numeros = [12378, 400, 325, 20, 21]
media, desvio = media_e_desvio(numeros)
print(f"média:{media}")
print(f"desvio padrão: {desvio}")
```

47. Contar Números em Lista: Conte quantas vezes cada número aparece em uma lista.


```python
def contar_numeros(lista):
  frequencia = {}
  for n in lista:
    if n in frequencia:
      frequencia[n] += 1
    else:
      frequencia[n] = 1
  return frequencia
numeros = list(map(int, input("Digite números separados por espaço: ").split()))
resultado = contar_numeros(numeros)

print("Frequência de cada número:")
for numero, qtd in resultado.items():
  if qtd == 1:
    print(f"{numero} aparece {qtd} vez")
  else:
    print(f"{numero} aparece {qtd} vezes")
```

48. Ordenação por Inserção: Implemente o algoritmo de ordenação por inserção (Insertion Sort).


```python
def insertion_sort(lista):
  for i in range(1, len(lista)):
    atual = lista[i]
    j = i - 1

  while j >= 0 and lista[j] > atual:
    lista[j + 1] = lista[j]
    j -= 1

  lista[j + 1] = atual

  return lista
numeros = list(map(int, input("digite números separados por espaço: ").split()))
ordenada = insertion_sort(numeros)
print(f"lista ordenada: {ordenada}")
```

49. Soma dos Dígitos: Calcule a soma dos dígitos de um número inteiro fornecido.


```python
def soma_digitos(numero):
  numero = abs(numero)
  soma = 0
  for digito in str(numero):
    soma += int(digito)
  return soma
n = int(input("Digite um número inteiro: "))
print(f"soma dos dígitos: {soma_digitos(n)}")
```
