22. Criptografia de Cifra de César: Crie uma função para criptografar uma mensagem usando a cifra de
César.

```python
mensagem = input("Digite a mensagem para criptografar: ") # pelo que pesquisei sempre tem uma chave mas não sei se é necessário pedir, então vou deixar ela já definida aqui em 2
tamanho = len(mensagem)
chave = 2
base = 'abcdefghijklmnopqrstuvwxyz' # precisa do alfabeto pra usar de base pra mudança
resultado = ''

for i in range(tamanho): # pegando com base no tamanho da entrada
    if mensagem[i] in base:
        posicao = base.index(mensagem[i]) # mover o index com base na posição atual
        nova_posicao = (posicao + chave) % 26 # 26 das letras do alfabeto, mudando a posicao com base na chave definida
        resultado += base[nova_posicao]
    else:
        resultado += mensagem[i] # soma as palavras porque se nao fica tudo junto

print(f"Mensagem criptografada: {resultado}")

```

23. Descriptografar Cifra de César: Crie uma função para descriptografar uma mensagem cifrada com a
cifra de César.

```python
mensagem = input("Digite a mensagem para descriptografar: ")
base = 'abcdefghijklmnopqrstuvwxyz' # precisa da base de novo
resultado = ''

for i in range(len(mensagem)): # aqui é a mesma lógica mas eu deixei pra pegar o tamanho direto aqui e ver se funcionava
    if mensagem[i] in base: # indo por letra
        posicao = base.index(mensagem[i]) # pegando a posicao da letra no alfabeto
        nova_posicao = (posicao - chave) % 26 # agora é subtração pra voltar
        resultado += base[nova_posicao]
    else:
        resultado += mensagem[i] # pra separar as palavras, novamente

print(f"mensagem descriptograda {resultado}")
```

24. Validação de CPF: Implemente uma função para validar um CPF brasileiro.

```python
cpf = input("Digite seu cpf: ") # pedindo cpf pro usuario
# 11 caracteres, sem letras, nao pode ser totalmente igual

if len(cpf) != 11 or not cpf.isdigit() or cpf == cpf[0] * 11: # verificando se tem 11, se os 11 são todos iguais e se são todos numeros
    print("Não é cpf brasileiro")
else:
  print(f"cpf verificado!")
```

25. Classificação de Números: Classifique uma lista de números em três categorias: positivos,
negativos e zeros.

```python
numList = [12378, -400, 325, 0, 20, -21, 0, 19, -6070, 801, 0, 1000, -2, 20] # lista pra facilitar

positivos = []
negativos = []
zeros = []

for i in numList: # percorrendo a lista
  if i > 0:
    positivos.append(i)
  elif i == 0:
    zeros.append(i)
  else:
    negativos.append(i)

print(f"positivos: {positivos} \nnegativos: {negativos} \nzeros: {zeros}")
```
