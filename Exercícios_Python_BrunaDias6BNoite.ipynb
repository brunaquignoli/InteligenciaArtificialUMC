6. Números Pares: Faça um programa que exiba todos os números pares de 1 a 20.

```python
pares = [] # achei melhor fazer uma lista pra ele exibir mais bonitinho em uma linha e não várias!

for i in range (1, 20):
  if (i % 2 == 0):
    pares.append(i)

print(f"números pares entre 1 e 20: {pares}") # usando o format novamente porque acho que fica mais bonito e porque uso muito javascript também, então sou acostumada em {}
```
