21. Busca Binária: Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

Aqui eu implementei a busca binária do jeito tradicional: peço o número que quero procurar e vou cortando a lista ao meio a cada tentativa, comparando com o valor do meio pra decidir se sigo pra esquerda ou pra direita, até achar ou a lista acabar. Reaproveitei a mesma lista de sempre pra facilitar.

```python
lista = [12378, 400, 325, 20, 21, 19, 6070, 801, 1000, 2, 20]
alvo = int(input("Digite o número que deseja procurar: "))

inicio = 0
fim = len(lista) - 1
meio = 0
achou = False

while inicio <= fim:
    meio = (inicio + fim) // 2

    if lista[meio] == alvo:
        achou = True
        break
    elif lista[meio] < alvo:
        inicio = meio + 1
    else:
        fim = meio - 1

if achou:
    print(f"O número {alvo} foi encontrado na posição {meio}.")
else:
    print(f"O número {alvo} não foi encontrado.")
```

22. Criptografia de Cifra de César: Crie uma função para criptografar uma mensagem usando a cifra de
César.

Pelo que pesquisei, sempre tem uma chave, mas não sei se era necessário pedir pro usuário, então deixei ela já definida em 2. Preciso do alfabeto pra usar de base pra mudança, e vou pegando com base no tamanho da entrada, movendo o índice com base na posição atual: uso o 26 das letras do alfabeto pra mudar a posição com base na chave definida. E vou somando os espaços também, porque senão as palavras ficam tudo junto.

```python
mensagem = input("Digite a mensagem para criptografar: ")
tamanho = len(mensagem)
chave = 2
base = 'abcdefghijklmnopqrstuvwxyz'
resultado = ''

for i in range(tamanho):
    if mensagem[i] in base:
        posicao = base.index(mensagem[i])
        nova_posicao = (posicao + chave) % 26
        resultado += base[nova_posicao]
    else:
        resultado += mensagem[i]

print(f"Mensagem criptografada: {resultado}")
```

30. Jogo da Forca: Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.

Deixei a palavra já fixa (spongebob, porque simmm) e criei uma lista pra guardar as letras que a pessoa já acertou. A cada rodada mostro a palavra com x no lugar das letras que ainda não foram descobertas, dou 6 tentativas, e vou tirando uma tentativa toda vez que a letra chutada não está na palavra. Quando todas as letras já tiverem sido descobertas, a pessoa vence.

```python
palavra = "spongebob"
letras_descobertas = []
tentativas = 6

while tentativas > 0:
    print("\npalavra:", end=" ")
    for letra in palavra:
        if letra in letras_descobertas:
            print(letra, end=" ")
        else:
            print("x", end=" ")
    print()

    chute = input("Digite uma letra: ").lower()

    if chute in letras_descobertas:
        print("já foi")
        continue

    if chute in palavra:
        letras_descobertas.append(chute)
        print("Acertou!!")
    else:
        tentativas -= 1
        print(f"errou! Tentativas restantes: {tentativas}")

    if all(letra in letras_descobertas for letra in palavra):
        print(f"Certo! a palavra é {palavra}")
        print("Você venceu!")
        break

if tentativas == 0 and not all(letra in letras_descobertas for letra in palavra):
    print(f"Você perdeu! a palavra era: {palavra}")
```

31. Ordenação por Bolha: Implemente o algoritmo de ordenação por bolha (Bubble Sort).

Implementei o bubble sort do jeito tradicional: vou comparando os números vizinhos da lista e trocando de lugar quando o da esquerda é maior que o da direita, repetindo isso várias vezes até a lista ficar toda ordenada. Usei a mesma lista de sempre.

```python
numeros = [12378, 400, 325, 20, 21, 19, 6070, 801, 1000, 2, 20]
n = len(numeros)

for i in range(n):
    for j in range(0, n - i - 1):
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]

print(f"Lista ordenada:  {numeros}")
```

37. Funções Recursivas: Implemente uma função recursiva para calcular o fatorial de um
número.

Professor, eu não usei recursão de verdade aqui, só refiz a mesma lógica do fatorial de antes, só que dentro de uma função, porque ainda não me sinto muito confortável com recursão. Mas o resultado final é o mesmo fatorial bonitinho com a explicação da conta.

```python
number = int(input("de qual número devemos calcular fatorial? "))

def fatorial(num):
    resultado = 1
    formatacao = " "

    for i in range(1, num + 1):
        resultado *= i

    formatNum = num
    for i in range(1, formatNum + 1):
        if formatNum != 1:
            formatacao += f" {formatNum} *"
            formatNum -= 1
        else:
            formatacao += f" {formatNum}"

    return resultado, formatacao.strip()

resultado, formatacao = fatorial(number)

print(f"O fatorial de {number} é {resultado} a partir de:\n{number}! = {formatacao}")
```

```python
n = int(input("Insira um número: "))

def primos_ate(n):
  primos = []
  for numero in range(2, n + 1):
      primo = True
      for divisor in range(2, numero):
          if numero % divisor == 0:
              primo = False
              break
      if primo:
          primos.append(numero)
  return primos

print(f"os primos são: {primos_ate(n)}" )
```

46. Calcular Média e Desvio Padrão: Calcule a média e o desvio padrão de uma lista de números.

Fiz uma função que calcula a média normal e depois usa ela pra calcular a variância (a diferença de cada número pra média, elevada ao quadrado) e por fim tira a raiz quadrada da variância pra achar o desvio padrão.

```python
def media_e_desvio(lista):
    media = sum(lista) / len(lista)
    variancia = 0

    for n in lista:
        variancia += (n - media) ** 2

    variancia /= len(lista)
    desvio = variancia ** 0.5
    return media, desvio

numeros = [12378, 400, 325, 20, 21]
media, desvio = media_e_desvio(numeros)
print(f"média:{media}")
print(f"desvio padrão: {desvio}")
```

48. Ordenação por Inserção: Implemente o algoritmo de ordenação por inserção (Insertion Sort).

Implementei o insertion sort (pesquisei esse aqui) pegando cada número da lista e comparando com os anteriores, empurrando os maiores pra frente até achar o lugar certo pra ele encaixar.

```python
def insertion_sort(lista):
  for i in range(1, len(lista)):
    atual = lista[i]
    j = i - 1

  while j >= 0 and lista[j] > atual:
    lista[j + 1] = lista[j]
    j -= 1

  lista[j + 1] = atual

  return lista
numeros = list(map(int, input("digite números separados por espaço: ").split()))
ordenada = insertion_sort(numeros)
print(f"lista ordenada: {ordenada}")
```

50. Cálculo de Frequência de Letras: Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

Criei uma função que deixa a frase em minúsculo e vai passando por cada caractere, contando só os que são letras (com isalpha()) num dicionário de frequência, e no final mostro tudo ordenado em ordem alfabética.

```python
def frequencia_letras(texto):
  freq = {}
  texto = texto.lower()

  for ch in texto:
    if ch.isalpha():
      if ch in freq:
        freq[ch] += 1
      else:
        freq[ch] = 1

  return freq
frase = input("Digite uma frase: ")
resultado = frequencia_letras(frase)

print("Frequência das letras:")
for letra, qtd in sorted(resultado.items()):
  print(letra, ":", qtd)
```
