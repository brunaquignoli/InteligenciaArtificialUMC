10. Sequência de Fibonacci: Imprima os primeiros 10 números da sequência de Fibonacci.

```python
fibonacci = []

inicio, num1 = 0, 1 # não sei se poderia fazer assim ou se deveria ser 100% tudo dentro do loop..
fibonacci.extend([inicio, num1]) #tive que pesquisar pra lembrar desse extend mas fica melhor!

for i in range(8): # aqui estava 10 mas deixei 8 já que já estou mostrando 2 números fora do loop
   num2 = inicio + num1
   fibonacci.append(num2)
   inicio = num1 # mudando eles pra usar sempre esse aqui e não precisar criar outras variáveis
   num1 = num2

print(f"os números da sequência de fibonacci são: {fibonacci}")
```

12. Contagem de Caracteres: Conte o número de ocorrências de cada caractere em uma string
fornecida.

```python
from IPython.core.completer import keyword
entrada = (input("Insira uma string: ")) # pedindo string pro usuario
entrada = list(entrada) # fazendo isso pra separar as letras

conta = {}
for i in entrada:
  if i in conta:
    conta[i] += 1 # contando se tem mais de uma
  else:
    conta[i] = 1 # deixando uma só na contagem

print(f"a string escolhida foi: {entrada}") # mostrando novamente a string escolhida
for letra, quantidade in conta.items(): # fazendo um for pra conseguir mostrar de maneira bonitinha
  print(f"a letra {letra} aparece {quantidade} vezes") # mostrando os resultados por letra

```
